# Employee Workforce and Salary Analysis

Course case study exploring team composition, positions, age groups, salary expenditure, and the relationship between age and salary. The workbook uses basketball-team and player-position labels; it should not be described as verified real-company HR data.

**Input:** place the course workbook `ABC Company.xlsx` in this folder. See `data/README.md` for missing dataset details.

**Method notes:** this exercise drops rows with missing values, so findings describe the retained sample. Height values are simulated for the exercise and are not evidence about actual height. Percentage and age-bin logic have been corrected. Outputs affected by those changes are cleared pending a fresh run with the original workbook. Other saved outputs are from the earlier run.


```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_excel("ABC Company.xlsx")
df = df.dropna()
df.head(10)
```

```python
df.info()
```

● Correct the data in the "height" column by replacing it with random
numbers between 150 and 180. Ensure data consistency and integrity
before proceeding with analysis.

```python
# Simulated heights required by the course exercise; not observed measurements.
rng = np.random.default_rng(42)
df["Height"] = rng.integers(150, 181, size=len(df))
df
```

```python
team_height = df.groupby("Team")["Height"].mean()
print(team_height.idxmax())
```

### Analysis Tasks:

● Determine the distribution of employees across each team and calculate
the percentage split relative to the total number of employees. 

```python
Team_num = df["Team"].value_counts()
Team_num
```

```python
Team_pec = Team_num.div(len(df)).mul(100)
Team_pec
```

```python
summary = pd.DataFrame({
    "Employee Count": Team_num,
    "Percentage (%)": Team_pec.round(2)
})
summary
```

```python
plt.figure(figsize=(10,6))
Team_num.head(10).plot(kind='bar', color='skyblue')
plt.title("Top 10 Teams by Employee Count")
plt.xlabel("Team")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### ANS Largest team :New Orleans Pelicans

● Segregate employees based on their positions within the company.


```python
position_count = df["Position"].value_counts()
print(position_count)
```

```python
plt.figure(figsize=(8,5))
position_count.plot(kind="bar", color="orange")

plt.title("Employees by Position")
plt.xlabel("Position")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

 ### ANS:Most common position is SG

● Identify the predominant age group among employees.

```python
df.columns
```

```python
c=df["Age"].max()
d=df["Age"].min()
print(c)
print(d)
```

```python
bins = [-np.inf, 19, 25, 30, 35, np.inf]
labels = ["Under 20", "20-25", "26-30", "31-35", "36+"]
df["Age group"] = pd.cut(df["Age"], bins=bins, labels=labels)
```

```python
age_group=df["Age group"].value_counts()
age_group
```

```python
print("Most common age group:")
print(age_group.idxmax())
```

```python
plt.figure(figsize=(8,6))
age_group.plot(kind="bar")
plt.title("Employee Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Employees")
plt.show()
```

● Discover which team and position have the highest salary expenditure.

```python
df.columns
```

```python
teamu=df.groupby("Team")["Salary"].sum()
print("Salary Expenditure by Team:")
print(teamu)
print("Team with Highest Salary Expenditure:")
print(teamu.idxmax())
```

```python
plt.figure(figsize=(10,6))

teamu.plot(kind="bar")

plt.title("Salary Expenditure by Team")
plt.xlabel("Team")
plt.ylabel("Total Salary")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### ANS:Highest salary expenditure team=Los Angeles Clippers

```python
posu=df.groupby("Position")["Salary"].sum()
print("Salary Expenditure by Position:")
print(posu)
print("Position with Highest Salary Expenditure:")
print(posu.idxmax())
```

```python
plt.figure(figsize=(10,6))

posu.plot(kind="bar", color="red")

plt.title("Salary Expenditure by Position")
plt.xlabel("Position")
plt.ylabel("Total Salary")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
